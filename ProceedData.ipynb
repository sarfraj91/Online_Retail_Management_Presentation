1.Handling Missing Values:

```python
missing_count = rawData.isnull().sum()

missing_percentage = (missing_count / len(rawData)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
rawData.isnull().sum()[rawData.isnull().sum() > 0]
```

```python
missing_desc = rawData[rawData["Description"].isnull()]

missing_desc[["StockCode", "Description", "Quantity", "UnitPrice", "Country"]].head(20)
```

```python
missing_desc["StockCode"].nunique()
```

```python
description_check = rawData[
    rawData["StockCode"].isin(missing_desc["StockCode"]) &
    rawData["Description"].notnull()
]

description_check[["StockCode", "Description"]].drop_duplicates().head(20)
```

```python
missing_description = rawData[rawData["Description"].isna()]

missing_description[["StockCode", "Description"]].head(20)
```

```python
description_mapping = (
    rawData.dropna(subset=["Description"])
           .drop_duplicates("StockCode")
           .set_index("StockCode")["Description"]
)

description_mapping.head()
```

```python
rawData["Description"] = rawData["Description"].fillna(
    rawData["StockCode"].map(description_mapping)
)
```

```python
rawData.isnull().sum()
```

There were initially 1,454 missing values in the Description column (0.27%). I used the StockCode to find matching product descriptions from other records and successfully recovered 1,342 values. The remaining 112 values could not be reliably identified, so I kept them as missing instead of guessing or adding incorrect information. These records were retained because other transaction information such as quantity, price, date, and country is still available.

```python
missing_customer = rawData[rawData["CustomerID"].isna()]

missing_customer.head(10)
```

```python
rawData["CustomerID"].isnull().sum()
```

CustomerID — Missing Values
Issue: There are 135,080 missing values in the CustomerID column, which is 24.93% of the dataset.

Action: I kept these values as missing (NaN) instead of filling or deleting them.

Why: CustomerID is needed to identify individual customers, but the other transaction details are still available. Therefore, these records can still be used for overall sales, product, and country analysis.

How handled: For customer-specific analysis, I will use only records where CustomerID is available.

Risk if not handled: Filling the missing values with an incorrect ID could create false customer information, while deleting the records would remove a large portion of useful transaction data.

```python
rawData[rawData.duplicated(keep=False)].head(20)
```

```python
for col in rawData.columns:
    print(f"\n--- {col} ---")
    print("Duplicate values:", rawData[col].duplicated().sum())
```

I distinguished between repeated values and duplicate records. Repeated values in individual columns are normal in transaction data. For example, one invoice can contain multiple products and one customer can make multiple purchases. Therefore, I only considered removing exact duplicate rows where the complete record was identical.

5.Creating calculated fields 


```python
rawData["Revenue"] = rawData["Quantity"] * rawData["UnitPrice"] 
rawData["CancellationFlag"] = rawData["InvoiceNo"].astype(str).str.startswith("C")
rawData["Month"] = rawData["InvoiceDate"].dt.to_period("M").astype(str)
rawData["Year"] = rawData["InvoiceDate"].dt.year
rawData.dtypes
```

```python
rawData.head(10)
```

```python
negative_quantity = (rawData["Quantity"] < 0).sum()

print("Negative Quantity rows:", negative_quantity)
```

5.Categorizing data 


```python
rawData["SalesType"] = np.where(
    rawData["CancellationFlag"] == True,
    "Cancellation/Return",
    "Sale"
)
```

Outliers

```python
Q1 = rawData["UnitPrice"].quantile(0.25)
Q3 = rawData["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

price_outliers = rawData[
    (rawData["UnitPrice"] < lower_bound) |
    (rawData["UnitPrice"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("UnitPrice outliers:", len(price_outliers))
```

```python
rawData[["StockCode", "Description", "Quantity", "UnitPrice", "Revenue"]].sort_values(
    "UnitPrice", ascending=False
).head(20)
```

Insights


```python
# 1. Monthly revenue
monthlyRevenue = (
    rawData.groupby("Month")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

# 2. Top products by revenue
topProducts = (
    rawData.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# 3. Top customers by revenue
topCustomers = (
    rawData.dropna(subset=["CustomerID"])
    .groupby("CustomerID")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# 4. Top countries by revenue
topCountries = (
    rawData.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

# 5. Cancellation analysis
cancellationSummary = rawData.groupby("SalesType").agg(
    Transactions=("InvoiceNo", "count"),
    Revenue=("Revenue", "sum")
)

print("===== TOP MONTHS =====")
print(monthlyRevenue.head(5))

print("\n===== TOP PRODUCTS =====")
print(topProducts)

print("\n===== TOP CUSTOMERS =====")
print(topCustomers)

print("\n===== TOP COUNTRIES =====")
print(topCountries)

print("\n===== CANCELLATION SUMMARY =====")
print(cancellationSummary)
```
